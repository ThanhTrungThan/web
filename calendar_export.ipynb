# Packages import

```python
import uuid
import datetime
from icalendar import Calendar, Event, vDatetime
import pandas as pd
import os
```

# Calendar export

```python
gr = input('enter your group name: ').strip()
csv_path = f"schedules/{gr}.csv"

if os.path.exists(csv_path):
    classes = pd.read_csv(csv_path, dtype=str)
    print(f'classes found: {len(classes)}')
else:
    pass
```

```python
calendar = Calendar()

calendar["dtstart"] = "TZID=Europe/Warsaw:20260223T000000"
calendar["dtend"] = "TZID=Europe/Warsaw:20260930T235959"
calendar["summary"] = f"Schedule for {gr} group"
calendar["PRODID"] = "-//UEK//Apollo to iCalendar export//EN"
calendar["version"] = "2.0"
```

```python
for index,row in classes.iterrows():
    event = Event()
    event["dtstart"] = row["Termin"].replace("-", "")+"T"+row["Start time"].replace(":", "")+"00"
    event["dtend"] = row["Termin"].replace("-", "")+"T"+row["End time"].replace(":", "")+"00"
    event["summary"] = row["Przedmiot"]
    event["location"] = row["Sala"]
    event["description"] = row["Nauczyciel"]
    event["UID"] = uuid.uuid4()
    event["dtstamp"] = vDatetime(datetime.datetime.now())
    calendar.add_component(event)
```

```python
classes['Sala'] = classes['Sala'].str.replace(
    r"(lab\.).*",
    r"\1",
    regex=True
)
```

```python
if not os.path.exists("./calendar"):
    os.mkdir("./calendar")
```

```python
with open(f"./calendar/{gr}.ics", "wb") as file:
    file.write(calendar.to_ical())
```
